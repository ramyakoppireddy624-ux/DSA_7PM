## **Bar chart automate**

```python
cat = visa_df.select_dtypes(include='object').columns
cat
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()    
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    plt.bar(keys,values)
    plt.xlabel(i)
    plt.ylabel('No of applicants')
    plt.title(f'{i}')
    con_df=pd.DataFrame(zip(keys,values),columns=[i,'No of applicants'])
 # step-3:
    foldername='bar charts'
    os.makedirs(foldername,exist_ok=True)
    cwd=os.getcwd()   
    new_dir=os.path.join(cwd,folder_name)
    filename=f'{i}.jpg'
    path=os.path.join(foldername,filename)
# step-4:
    plt.savefig(path)
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.pie(values,
            explode=[0.05 for i in range(len(keys))],
            labels=keys,
            autopct='%0.2f%%',
            startangle=30,
            radius=1) 
    con_df=pd.DataFrame(zip(keys,values),columns=[i,'No of applicants'])
    folder_name='pie_charts'
    os.makedirs(folder_name,exist_ok=True)
    cwd=os.getcwd()   
    new_dir=os.path.join(cwd,folder_name)
    filename=f'{i}.jpg'
    path=os.path.join(folder_name,filename)
# step-4:
    plt.savefig(path) 
    plt.show()
```
